# NumPy

## Vector - Funcionalidades y usos.

```python
# Importacion de la biblioteca NumPy
import numpy as np
```

### Creacion de arrays

```python
# Para crear un arraay, nombramos la variables y le pasamos el sig. comando.

# Esto crea un vector 3x1 (en el caso de visualizarlo)
arr_01 = np.array([1,2,3])
print(f'Este es mi primer arraya {arr_01}')

# Tambien podemos usar crear arrays como autocompletado usando metodos.
arr_zeros = np.zeros(3)
print(f'Este es un array formado con el metodo zeros {arr_zeros}')

arr_ones = np.ones(3)
print(f'Este es un array formado con el metodo ones {arr_ones}')

arr_values_random = np.random.random(3)
print(f'Este es un array formado con el metodo random {arr_values_random}')
```

### Operaciones aritmeticas

```python
# Antes de iniciar con la operaciones aritmeticas, voy a crear los arrays que
# usare para estas operaciones.
arr_data_01 = np.array([1,2])
arr_ones_01 = np.ones(2)
print(f'Estos son los valores de los arrays {arr_data_01}, {arr_ones_01}')

#Suma
arr_suma = arr_data_01 + arr_ones_01
print(f'La suma de los arrays es: {arr_suma}')

#Resta
arr_resta = arr_data_01 - arr_ones_01
print(f'La resta de los arrays es: {arr_resta}')

#Multiplicacion
arr_mul = arr_data_01 * arr_ones_01
print(f'La multiplicacion de los arrays es: {arr_mul}')

#Division
arr_div = arr_data_01 / arr_ones_01
print(f'La division de los arrays es: {arr_div}')

#Tambien se puede multiplicar los arrays por es escalar.
arr_mul_k = arr_data_01 * 2.5
print(f'La multiplicacion del array por el escalar 2.5 es: {arr_mul_k}')
```

### Indexacion en un array

```python
# Podemos indexar o cortar un array de las formas que queramos
arr_data_02 = np.array([1,2,3])

# El primer elemento
print(f'Este es el primer elemento: {arr_data_02[0]}')

# El segundo elemento
print(f'Este es el segundo elemento: {arr_data_02[1]}')

# Cortar varios elementos
print(f'Estos son los dos primeros elementos: {arr_data_02[0:2]}')

# Cortar varios elementos
print(f'Estos son los dos ultimos elementos: {arr_data_02[1:]}')
```

### Algunos Metodos de los Arrays

```python
from typing_extensions import MappingView
#Algunos metodos que podemos encontrar dentro de NumPy

#Suma todos los valores del array.
suma_arr = arr_data_02.sum()
print(f'Suma: {suma_arr}')

#Devuelve el menor valor de array.
min_arr = arr_data_02.min()
print(f'Minimo: {min_arr}')

#Devuelve el mayor valor de array.
max_arr = arr_data_02.max()
print(f'Maximo: {max_arr}')

#Devuelve el promedio de valores de array.
mean_arr = arr_data_02.mean()
print(f'Maximo: {mean_arr}')

#Esos son algunos de las metodos que pondes llamar con NumPy.
```

## Matrices - Funcionalidades y usos.

Una matriz puede entenderse conceptualmente como una coleccion de vectores organizados en una estructura bidimensional.

### Creacion de una matriz con NumPy

```python
# Utilizamos el mismo metodo, solo que ahora es una lista de una lista.
matriz_01 = np.array([[1,2], [3,4]])
print(f'{matriz_01}')

#Esta matriz tiene la forma 2x2 ((m,n) donde m son las filas y n las columnas).
print(f'La forma de la matriz es {matriz_01.shape}')

# Tambien podemos crear matrices con los metodos, one, zero, random etc.

matriz_one = np.ones((3,4))
matriz_cero = np.zeros((2,3))

print(f'Esta es una matriz con unos {matriz_one}')
print(f'Esta es una matriz con ceros {matriz_cero}')
```

### Operaciones aritmeticas

Podemos realizar todas las operaciones aritmeticas si las matrices son de la misma forma (tamanio).

```python
matriz_01 = np.array([[1,2], [3,4]])
matriz_02 = np.array([[2,2], [2,2]])

#Suma de matrices
matriz_suma = matriz_01 + matriz_02
print(matriz_suma)

#Suma de resta
matriz_resta = matriz_01 - matriz_02
print(matriz_resta)

#Multiplicaciones
matriz_mul = matriz_01 * matriz_02
print(matriz_mul)

#Division
matriz_div = matriz_01 / matriz_02
print(matriz_div)
```

Existe una forma de realizar operaciones algebricas si las matrices son distintas, la condicion es que una de las matrices sea un vector igual en el numero de filas o columnas que la otra matriz.

```python
matriz_random = np.random.random((3,3))
vector_random = np.random.random(3)

print(f'{matriz_random}\n')
print(vector_random)

# Aca utilizamo broadcasting que permite operar entre arrays de formas
# diferentes estirando virtualmente el mas chico para que encaje con el mas grande.
sum_matriz_vector = matriz_random + vector_random
print(f'\n \n Suma: \n{sum_matriz_vector}')
```

### Dot Product
Es una funcion que multiplica una matriz por otra de distinto tamanio, para que se puede realizar esta funcion se tiene que cumplir con la siguiente condicion, la cantidad de columnas de la primera matriz debe ser igual a la cantidad de filas de la segunda matriz. Esto da como resultado una matriz de forma cantidad de filas de la primera matriz y cantidad de columnas de la segunda. Ejemplo, la primera matriz (2,3) y la segunda (3,4) igual a otra matriz (2,4).

```python
# Creamos las matrices que vamos a usar para el ejemplo.
matriz_dot01 = np.array([[1,2,3]])
matriz_dot02 = np.array([[1,10],[100,1000],[10000,100000]])

# Como observamos creamos una matriz de 1x3 y una de 3x2.
# estas cumplen con la condicion para utilizar la funcion dot.
print(f'Forma de la matriz 01: {matriz_dot01.shape}')
print(f'Forma de la matriz 02: {matriz_dot02.shape}')

# Utilizamos la funcion dot.
# Esta multiplica y crea un matriz  1x2
result_arr = matriz_dot01.dot(matriz_dot02)

print(f'\n Matriz: \n {result_arr}')
```

## Indexacion de Matrices

Para encontrar la ubicacion de una matriz, diremos que el indice se compone por
el primer elemento (x) [x,y] el numero de fila que ocupa y el segundo elemento
seniala (y) el numero de columna que ocupa.

```python
# Creamos una matriz de forma (3x2)
matriz_indice = np.array([[1,2],[3,4],[5,6]])

# Mostramos su forma y la matriz.
print(f'Forma de la matriz: {matriz_indice.shape}\n')
print(f'Matriz: \n{matriz_indice}')

# Seleccionamos el elemento este en la ubicacion [0,1]
# No olvidemos que el primero numero es la ubicacion de la fila
# y el segundo es el numero de la columna.
print(f'\nElemento en la ubicacion [0,1] \nEs el:{matriz_indice[0,1]}')

# En esta ocacion seleccionamos varios elementos [1:3] como dentro de los parentesis
# no hay una coma, estamos haciendo referencia a las filas  completa desde el lugar
# 1 hasta el 3, estos nos devolvera todos los elementos de dichas filas sin tener en
# cuenta las columnas.
print(f'\nElemento en la ubicacion [1:3] \nEs el:\n{matriz_indice[1:3]}')

# Otro caso, ahroa selecionamos las filas a tener en cuenta y tambien las columnas
# [0:2,0] nos devolvera los elementos de las filas 1 y 2 tomando solo los elementos
# de la columna 1, no nos olvidemos que empezamos a contar de 1 cuando el indece en 0
print(f'\nElemento en la ubicacion [0:2,0] \nEs el:\n{matriz_indice[0:2,0]}')
```

### Algunos metodos que podemos usar en las Matrices

```python
# Utilizaremos la matriz del ejemplo de arriba0
matriz_metodos = matriz_indice

# Devuelve el valor maximo de la matriz
valor_max = matriz_metodos.max()
print(f'El valor maximo es: {valor_max}')

# Devuelve el valor minimo de la matriz
valor_min = matriz_metodos.min()
print(f'El valor minimo es: {valor_min}')

# Devuelve la suma los valores de la matriz
valor_sum = matriz_metodos.sum()
print(f'La suma de todos los elementos de la matriz es: {valor_sum}')
```

No solo podemos agregar todos los valores de una matriz, sino que tambien podemos agregar a través de las filas o columnas usando el parametro axis.

```python
#suma todos los elementos por columna y nos devuelve el valor por cada una.
suma_colum = matriz_metodos.sum(axis=0)
print(f'La suma de las columnas es:{suma_colum}')

#suma todos los elementos por filas y nos devuelve el valor por cada una.
suma_filas = matriz_metodos.sum(axis=1)
print(f'La suma de las filas es:{suma_filas}')
```

### Transposing and Reshaping (Transponer y Remodelar)
Vamos a usar estos metodos cuando querramos hacer dot, algun tipo de operacion aritmetica y necesitemos que las matrices tengan la misma forma.

```python
#Transponer, utilizando el metodo .T
#Utilizaremos la matriz_metodos, del ejemplo anterior.
matriz_metodos_tranp = matriz_metodos.T
print(f'Matriz transpuesta: \n {matriz_metodos_tranp}')
```

```python
#Remodelar, utilizando el metodo .reshape
#creamos una matriz de la forma (6x1)
data_modelar = np.array([[1],[2],[3],[4],[5],[6]])
print(f'Matriz: \n{data_modelar}')
print(f'\nForma:{data_modelar.shape}')

#Utilazmos el metodo reshape
data_remodelar = data_modelar.reshape(2,3)

print(f'\nMatriz remodelada: \nMatriz: \n{data_remodelar}')
print(f'\nForma:{data_remodelar.shape}')

#Utilazmos el metodo reshape
data_remodelar = data_modelar.reshape(3,2)

print(f'\nMatriz remodelada: \nMatriz: \n{data_remodelar}')
print(f'\nForma:{data_remodelar.shape}')
```
